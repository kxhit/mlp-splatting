```python
from utils.read_write_model import *


def w2c_to_c2w(image):
    R = qvec2rotmat(image.qvec)  # shape (3, 3)
    t = image.tvec  # shape (3, )

    c2w = np.eye(4)
    c2w[:3, :3] = R.T
    c2w[:3, 3] = -R.T @ t
    return c2w
```
